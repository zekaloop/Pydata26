### 2. 시리즈 (Series)
* 객체 생성, 주요 속성 확인

```python
numbers = [1, 2, 3, 4, 5]
# 시리즈 생성
sr1 = pd.Series(numbers)
sr1
```

```python
# 속성 확인 : 데이터 타입, 행과 열의 수, 데이터 수, 이름
print(sr1.dtype, sr1.shape, sr1.size, sr1.name)

# 속성 확인 : 인덱스
print(sr1.index)

# 속성 확인 : 데이터 값
print(sr1.values)
```

```python
numbers = [1, 2, 3, 4, 5]
# 실수형으로 시리즈 생성
sr2 = pd.Series(numbers, dtype=float)
sr2
```

```python
chars = ['a', 'b', 'c', 'd', 'e']
# 시리즈 생성 시 index와 name 지정
sr3 = pd.Series(numbers, index=chars, name='MySeries')
sr3

# 속성 확인
sr3.dtype, sr3.name, sr3.index
```

```python
# 시리즈 생성 후에 index와 이름 추가하기
sr4 = pd.Series(numbers)
sr4.index = chars
sr4.name = 'MySeries'
sr4
```

```python
dict_data = {'a' : 1, 'b' : 2, 'c' : 3, 'd' : 4, 'e' : 5}
# dictionary로부터 시리즈 생성하기
dict_sr = pd.Series(dict_data)

# 속성 확인
print(dict_sr.index)
print(dict_sr.values)
    
dict_sr
```

### 3. 데이터프레임 객체 생성, 주요 속성 확인

### 4. 데이터프레임의 열 다루기 (추가, 삭제, 수정)

```python
score_df = pd.DataFrame(score_dict)
score_df
```

```python
# '사회' 컬럼 추가
scores =[85, 88, 92]
score_df['사회'] = scores
score_df
```

```python
# 모든 칼럼의 값을 더해서 '총점' 칼럼 만들어 추가하기
score_df['총점'] = score_df['국어'] + score_df['영어'] + score_df['수학'] + score_df['사회']
score_df
```

```python
# '평균' 칼럼 추가
score_df['평균'] = score_df['총점'] / 4
score_df
```
