## 4. 데이터형 확인과 변환

```python
mylist = [0, 1, 2, 3, 4, 5]
nda=np.array(mylist)

# nda 변수 데이터형 확인
print(type(mylist), type(nda))

# nda 요소의 데이터형 확인
nda.dtype
```

```python
nda2 = np.array([0, 1, 2, 3, 4, 5])
print(nda2, nda2.dtype)
```

```python
# 데이터형 변환 : astype()
a1 = np.array(['1.2', '2.3', '3.4', '4.5'])
print(a1, a1.dtype)

# 문자열 -> 실수형으로 변환
a2 = a1.astype(float)
print(a2, a2.dtype)

# 실수형 -> 정수형으로 변환
a3 = a2.astype(int)
print(a3, a3.dtype)


# 문자열 -> 정수형으로 변환
a5 = np.array(['1', '2', '3'])
print(a5, a5.dtype)
a4 = a5.astype(int)
print(a4, a4.dtype)
```

## 6. 배열의 인덱싱

```python
# 1차원 배열의 인덱싱
d = np.arange(1, 10)
print(d, d.shape)
print(d[0], d[1], d[2], d[-1], d[-2])

# index 0의 데이터를 10으로 변경
d[0] = 10
print(d)
```

```python
# 2차원 배열의 인덱싱
d2 = np.arange(1, 11).reshape(2, -1)

print(d2[0][0]) # 배열명[행][열] 형태
print(d2[0,0]) # 배열명[행,열] 형태
print(d2[1,1])
print(d2[1,-1])
```

```python
# 3차원 배열의 인덱싱
d3 = np.arange(12).reshape(2, 2, 3)
print(d3)

print(d3[0][0][0]) # 배열명[면][행][열] 형태
print(d3[0, 0, 0]) # 배열명[면,행,열] 형태
print(d3[1, 1, 2])
print(d3[-1, -1, -1])
```

## 7. 배열의 슬라이싱

```python
# 1차원 배열의 슬라이싱
d = np.arange(1, 10)
print(d)
print(d[2:3], d[:], d[:-1])
print(d[::2], d[::-1])
```

```python
# 2차원 배열의 슬라이싱
d2 = np.arange(1, 11).reshape(2, -1)
print(d2)

# 0행만 슬라이싱
print(d2[0])
# 0행의 전체 열 슬라이싱
print(d2[0])
# 마지막 열 제외한 전체 행과 열 슬라이싱
print(d2[:,:-1])
# 2행 2열 간격으로 전체 행과 열 슬라이싱
print(d2[::2, ::2])
```

```python
# 슬라이싱된 요소값을 한꺼번에 변경
print(d2[::2,::3])
d2[::2,::3]=30 
print(d2)
```

```python
# 0번 행 요소를 모두 50으로 변경
d2[0] = 50
print(d2)
```

```python
# 전체 요소값을 한꺼번에 변경
d2[:,:] = -1
print(d2)
```

```python
# 3차원 배열의 슬라이싱
d3 = np.arange(12).reshape(2, 2, 3)
print(d3)
print(d3[:,:-1,:-1])
print(d3[:,1:,1:])
print(d3[:,:,-1])
print(d3[:,-1,-1])
```

## 8. 불 인덱싱 (Boolean Indexing)

```python
data = np.array([4, 6, 7, 3, 2, 0])
print(data[data > 3])
```

```python
print(data > 3)
```

```python
cond = [False, True, True, False, False, True]
data[cond]
```

```python
print(data)
# 2보다 크고 6보다 작은 요소만 추출
print(data[(data > 2) & (data < 6)])
# 2로 나눈 나머지가 0인 요소만 추출
print(data[data % 2 == 0])

# 평균보다 작은 값만 추출
print(data[data < data.mean()])
```

## 10. 넘파이 배열의 정렬

```python
# 기본 정렬: sort 함수
numbers = np.array([2, 8, 9, 7, 3, 1, 4, 5, 6])
print(numbers, numbers.shape)

# 오름차순 정렬
print(np.sort(numbers))

# 내림차순 정렬
print(np.sort(numbers)[::-1])
```

```python
# 2차원 배열의 정렬
arr2d = np.array([[5, 6, 7, 8],
                [4, 3, 2, 1],
                [10, 9, 12, 11]])
print(arr2d)
print(arr2d.shape)
```

```python
np.sort(arr2d)
```

```python
# 열 방향으로 같은 행의 요소 정렬
np.sort(arr2d, axis=1)
```

```python
# 행 방향으로 같은 열의 요소 정렬
np.sort(arr2d, axis=0)
```

```python
# 인덱스 반환 정렬: argsort 함수
a=np.array([2, 8, 9, 7, 3, 1, 4, 5, 6])
print(a, a.shape)
sorted_indexes = np.argsort(numbers)
print(sorted_indexes)
```

#### [연습문제] 이름과 성적이 각각의 ndarray로 주어진 경우 이름을 성적 순서로 출력하기

```python
names = np.array(['John', 'Mike', 'Sarah', 'Kate', 'Samuel'])
scores= np.array([78, 95, 84, 98, 88])

sorted_index = np.argsort(scores)[::-1]

print('성적 내림차순 정렬 시 scores의 인덱스:', sorted_index)
print('성적 오름차순으로 names의 이름 출력:', names[sorted_index])
print('성적 내림차순으로 출력 : ', scores[sorted_index])
```
