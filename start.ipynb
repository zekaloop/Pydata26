```python
number_list = [23, 45, 27, 11, 25, 65, 78]

print(getIndex(number_list, 25))
print(getMax(number_list))
print(getMin(number_list))
print(countGT(number_list, 42))
print(sumList(number_list))
swapList(number_list)
print(number_list)
```

```python
def getIndex(num_list, target):
    for i in range(len(num_list)):
        if num_list[i] == target:
            return i
```

```python
def getMax(num_list):
    max_num = 0
    for i in num_list:
        if i > max_num:
            max_num = i
    return max_num
```

```python
def getMin(num_list):
    min_num = num_list[0]
    for i in num_list:
        if i < min_num:
            min_num = i
    return min_num
```

```python
def countGT(num_list, target):
    count=0
    for i in num_list:
        if int(i)>target:
            count+=1
    return count
```

```python
def sumList(num_list):
    total = 0
    for i in num_list:
        total += i
    return total
```

```python
def swapList(num_list):
    n = len(num_list)
    for i in range(n // 2):
        temp = num_list[i]
        num_list[i] = num_list[n - 1 - i]
        num_list[n - 1 - i] = temp
        
    return num_list
```
